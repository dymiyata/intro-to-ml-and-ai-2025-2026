```python
from sklearn.datasets import load_breast_cancer # thing to import data
from sklearn.model_selection import cross_val_score # thing to do cross validation
from sklearn.linear_model import LogisticRegression # thing to do logistic regression
```

### Doing 5-fold Cross Validation

```python
X, y = load_breast_cancer(return_X_y=True)
```

```python
model = LogisticRegression(max_iter=20000)
```

```python
scores = cross_val_score(model, X, y, cv=5, scoring="accuracy") # replaces "model.fit(X,y)" when doing cross validation
```

```python
print("scores:", scores)
print("Average:", scores.mean())
```

### Doing Regularization

Secretly Logistic Regression in sklearn was doing regularization all along (with $\alpha = 1$)

But it uses a hyperparameter $C$. Where you can think of it as:
$$ C = \frac{1}{\alpha} $$

```python
model_reg = LogisticRegression(max_iter=20000, C=100) # Setting the value of C changes the strength of regulariaziton
```

```python
scores_reg = cross_val_score(model_reg, X, y, cv=5, scoring="accuracy") # replaces "model.fit(X,y)" when doing cross validation
print("scores:", scores_reg)
print("Average:", scores_reg.mean())
```
